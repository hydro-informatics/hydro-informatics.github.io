````python
rho_k =  870. # kg / m3 - wood density
rho_f = 1000. # kg / m3 - water density
g = 9.81      # N / kg - gravitational acceleration
b_x = 1.5     # m - box width
b_y = 2.0     # m - box length
h = 1.5       # m - box height
w = 0.01      # m - wall thickness
````

Calcul de la force de poids $ F_g = \rho_k \cdot g \cdot V_k = \rho_k \cdot g \cdot \left[b_x \cdot b_y \cdot h - \left(h - w\right) \cdot  \left(b_x - 2 w\right) \cdot \left(b_y - 2 w\right)\right] $ :

````python
V_k = b_x * b_y * h - (b_x - 2 * w) * (b_y - 2 * w) * (h - w)
F_g = V_k * rho_k * g
print(f'Body volume V_k = {V_k:.2f} m3\nWeight force: {F_g:.2f} N')
````

Calcul du volume déplacé $ V_v = F_g / \left(\rho_f \cdot g \right) $:

````python
V_v = F_g / (rho_f * g)
print(f'Displaced volume: {V_v:.2f} m3')
````

Calcul du projet (profondeur d'immersion) $ d = V_v / \left(b_x \cdot b_y\right) $ et du franc-bord $ f = h - d $ :

````python
d = V_v / b_x * b_y
f = h - d
print(f'Draft d = {d:.2f} m\nFreeboard f = {f:.2f} m')
if f >= 0:
    print("The body floats.")
if f < 0:
    print("The body sinks.")
````

## Stabilité flottante

### Deuxième moment de la zone

Calcul du deuxième moment de la zone $I_0 = \frac{b_{\text{x}} ^3 \cdot b_{\text{y}} }{12}$ par rapport à 0:

````python
I_0 = b_x**3 * b_y / 12
print(f'Second moment of area I_0 = {I_0:.2f} m4')
````

### Elévations

Hauteur $h_g$ du centre de gravité:

````python
V_k_side_walls = 2 * (b_y - 2 * w) * (h - w) * w
V_k_front_back = 2 * b_x * (h - w) * w
V_k_bottom = b_x * b_y * w
h_g = ((V_k_side_walls + V_k_front_back) * (w + h / 2) + V_k_bottom * w / 2) / V_k
print(f'h_g = {h_g:.2f} m')
````

Hauteur $h_v$ du centre de flottabilité:

````python
h_v = d / 2
print(f'Height of the center of buoyancy h_v = {h_v:.2f} m')
````

Différence de hauteur $h_{gv}$ entre le centre de gravité et le centre de flottabilité:

````python
h_gv = h_g - h_v
print(f'Height difference h_gv = {h_gv:.2f} m')
````

Hauteur métacentrique $h_M$ et stabilité flottante :

````python
h_M = I_0 / V_v - h_gv
print(f'Metacentric height h_M = {h_M:.2f} m')

if h_M > 0:
    print("Stable equilibrium.")
elif h_M < 0:
    print("Unstable equilibrium.")
else:
    print("Neutral equilibrium.")
````
